### 5. Customer satisfaction towards category & product 
-----------------------------------------------------------------
### a. Which categories (top 10) are maximum rated & minimum rated? 


```python
t=cust_ord_item_payment_reviwe_product_seller_geo.groupby('product_category_name')['review_score'].mean().reset_index().sort_values('review_score',ascending=False,ignore_index=True).head(10)
t.rename(columns={'review_score':'Max_rated'},inplace=True)
g=cust_ord_item_payment_reviwe_product_seller_geo.groupby('product_category_name')['review_score'].mean().reset_index().sort_values('review_score',ascending=True,ignore_index=True).head(10)
g.rename(columns={'review_score':'Min_rated'},inplace=True)

p=pd.concat([t,g],axis=1)
p
```

### b. Which products (top10) are maximum rated & minimum rated? 



```python
t=cust_ord_item_payment_reviwe_product_seller_geo.groupby('product_id')['review_score'].mean().reset_index().sort_values('review_score',ascending=False,ignore_index=True).head(10)
g=cust_ord_item_payment_reviwe_product_seller_geo.groupby('product_id')['review_score'].mean().reset_index().sort_values('review_score',ascending=True,ignore_index=True).head(10)
t.rename(columns={'review_score':'Maximum_rated'},inplace=True)
g.rename(columns={'review_score':'Min_rated'},inplace=True)
p=pd.concat([t,g],axis=1)
p
```
